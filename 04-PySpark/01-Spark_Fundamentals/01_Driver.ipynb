**Driver**

1.  It is one of the most important component of Apache Spark.

2.  It is the brain of Spark Applications.

3.  It allowsthe user to use all the functionalities of Pyspark.

4.  It helps in creating the spark session and configures the spark context.

5.  It helps in processing the results through the clusters and get the actual result after the entire dataprocessing tasks.


```python
##### import all the libraries for spark session
from pyspark.sql           import SparkSession
from pyspark.sql.functions import col,sum
from pyspark.sql.types     import *
```

```python
##### Creating the spark session
session = SparkSession.builder.appName("Driver").getOrCreate()

##### get the version of spark session
print(session.version)
```

**OBSERVATIONS:**

1.  It is the point where our Driver works. This is because the driver creates the spark session.

```python
##### Create the dataframe with the help of spark session
data = [
    (101, "Amit", "IT", "Hyderabad", 60000),
    (102, "Ravi", "HR", "Bangalore", 50000),
    (103, "John", "IT", "Chennai", 70000),
    (104, "Neha", "Finance", "Delhi", 80000),
    (105, "Priya", "IT", "Hyderabad", 90000),
    (106, "Rahul", "HR", "Pune", 55000),
    (107, "Anita", "Finance", "Chennai", 75000),
    (108, "Arjun", "IT", "Bangalore", 85000),
    (109, "Sneha", "Sales", "Delhi", 65000),
    (110, "Vikas", "IT", "Pune", 72000)
]
columns = [
    "employee_id"         ,
    "employee_name"       ,
    "department"          ,
    "city"                ,
    "salary"
]

##### create the dataframe from it
df = session.createDataFrame(data,columns)

##### display the dataframe
df.show()
```

```python
##### perform the actual data transformations

##### filter all the employees who work in 'IT' department
it_employees=df.filter(df.department=='IT')

##### filter only name and salary of the IT department employees
result = it_employees.select(col("employee_name"),col("salary"))

result.show()
```

```python
##### stop the spark session
session.stop()
```

**Working mechanism of the above process**

1.  The driver  creates the spark session.

2.  The driver acts as an intermediate between the spark application and the cluster manager.

3.  With the help of spark session, the user can access all the functionalities of pyspark.

4.  Then for spark transformations, instead of executing it immediately, a logical plan has been made and this plan is stored in action.

5.  Then the action is executed. This process is called lazy evaluation.

6.  Then after action is executed, a job process is created and then is sent to the cluster manager.

7.  This cluster manager allocated the executor to each and every cluster. It even allocates the memory to every executor.

8. Then the data from the cluster manager is sent to the cluster.

9. Then in cluster, it has several worker nodes along with its executors.

10. The data gets divided among the worker modes with its executors.

11. Then each worker node performs the data processing tasks.

12. Each executor in the worker nodes executes the data and gets the processed result.

13. The processed result from each executor is combined to get a final result.

14. This result is then sent back to the driver program.

15. Thus in this way the data processing in the apache spark works.
