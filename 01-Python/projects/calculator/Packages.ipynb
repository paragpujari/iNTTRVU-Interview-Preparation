##### Package

1.  A Package can be defined as the collection of all the python modules inside the directory.

Package

│

├── module1.py

├── module2.py

└── module3.py


##### Advantages of Packages

1.  Packages help in reusing the code.

2.  It helps in organizing the code.

3.  It helps in grouping multiple number of similar modules together.

4.  It avoids writing huge python code.

5.  It maintains large applications very easily.
