##### OBSERVATIONS:

1.  Python is dynamically typed.

2.  The same variable can have different values during the different runtime execution of the progam.

3.  Based on the values during different runtime execution of the program, the data type also get determined with respect to the values.

```python
x = 10
print(x)
print(type(x))       ####### int


x = 80.87
print(x)
print(type(x))       ####### float

x = (2+3j)
print(x)
print(type(x))       ####### Complex

x = True
print(x)
print(type(x))       #######  Boolean
```

##### OBSERVATIONS:

1.  The variable x has four different values defined during different runtime execution of the program.

2.  As four different values get assigned to the same variable during different runtime execution of the program, so four different datatypes gets determined automatically.

##### Rules for defining the variable

(a.)    A variable must begin with an alphabet or an underscore

(b.)    It should not begin with a digit

(c.)    It must have an underscore in between if needed

(d.)    It should not be any keyword

(e.)    It should not have hyphen in between

(f.)    It should not used any special character like ($,%,&) etc.

```python
##### Rule 1: A variable must begin with an alphabet or an underscore

name = 'Parag'     ##### Variable begins with alphabet
_name = 'Parag'    ##### Variable begins with underscore

print(f"Variable begins with alphabet is {name}")
print(f"Variable begins with underscore is {_name}")
```

```python
##### Rule 2: It should not begin with a digit

#1name = 'Parag'
```

```python
##### Rule 3: It must have an underscore in between if needed

first_name = 'Parag'
last_name = 'Pujari'

print(first_name)
print(last_name)
```

```python
##### Rule 4: It should not be any keyword

#class = 'Parag'
```

```python
##### Rule 5: It should not have hyphen in between

#first-name = 'Parag'
```

```python
##### Rule 6: It should not used any special character like ($,%,&) etc

#$name = 'Parag'
#%name = 'Parag'
#&name = 'Parag'
```

```python
##### Python is case sensitive

age = 25
Age = 30

print(age)
print(Age)
```

##### OBSERVATIONS:

1.  As 'age' and 'Age' are defined differently and as Python is case sensitive, so Python treats 'age' and 'Age' differently.

2.  So age = 25 and Age = 30.

```python
##### Naming convention ---> It must have a snake case

first_name = 'Parag'
last_name  = 'Pujari'

print(first_name)
print(last_name)
```

40.  You have the following variables:

original_price = 1000

discount = 200

tax = 80

Calculate the final price using only variables and arithmetic.

The logic is:

Price after discount = original price - discount

Final price = price after discount + tax

```python
original_price = 1000

discount = 200

tax = 80

##### Calculate the price after discount
price_after_discount = original_price - discount
print(f'Price after discount is {price_after_discount}')

##### Calculate the final price
final_price = price_after_discount + tax
print(f'Final Price is {final_price}')
```
