##### Input/Output Operations

1.  Input --------------->   It is used to take the input from the user.

2.  Output -------------->   It is used to display the output onto the console.

```python
##### enter your name
name = input("Enter your name")
print(name)
```

##### OBSERVATIONS:

1. input() ----> It enters the value in string form.

```python
##### enter your age
age = input("Enter your age")
print(age)
```

##### Numeric Input

It considers the input in int/float form

```python
##### enter age in integer form
age = int(input("Enter your age"))
print(age)
```

```python
##### enter your salary in float form
salary = float(input("Enter your salary"))
print(salary)
```

##### Output

It prints the output value in console.

```python
print("Hello")
```

##### sep specifies what should appear between multiple values

```python
print("Python","SQL","Scala","Pyspark",sep="|")
```

##### OBSERVATIONS:

1.  "|" is printed between the values.

##### end

If end is not specified in print(), then print will move to the next line.

```python
print("Hello")
print("Python")
```

##### OBSERVATIONS:

1.  Here end is not specified in print, so two print statements will be printed line by line.

```python
print("Hello", end=" ")
print("Python")
```

##### OBSERVATIONS:

1.  Here end is specified in print with " ", so two print statements will be printed in the same line.
