```python
file = open("data.txt","a")
```

##### Reading operations

1. read()

It is used to read the entire file.

```python
##### open the file and read it

data = open("data.txt","r")

##### Read the contents in the file
x = data.read()

print(x)

##### close the file
data.close()
```

2. read(n)

It is used to read 'n' number of characters from the file.

```python
##### open and read 'n' no of contents
file = open('data.txt','r')

#### Read n number of contents inside the file
x = file.read(10)
print(x)

##### close the file
file.close()
```

3. readline()

It reads one line

```python
##### open and read the file
file = open('data.txt','r')

##### Read one line of file
x = file.readline()
print(x)

##### close the file
file.close()
```

4. Multiple readline()

It read multiple number of lines line by line

```python
##### open and read multiple number of lines
file = open("data.txt","r")

##### Read multiple lines line by line in a file
line1 = file.readline()
line2 = file.readline()

print(line1)
print(line2)

##### close the file
file.close()
```

5. readlines()

It reads all the lines

```python
##### open and read the file
file = open("data.txt","r")

##### read all lines
x = file.readlines()
print(x)

##### close the file
file.close()
```

##### Writing Operations

1. write()

It is used to write the contents into the file.

```python
##### open the file in write mode
file = open("data.txt","w")

##### write the contents into the file
x = file.write("Python File Handling")
print(x)

#### close the file
file.close()
```

2. write()

It can be used to write the contents for multiple lines.

```python
##### open the file in write mode
file = open("data.txt","w")

##### write contents in multiple line
file.write("Python\n")
file.write("SQL\n")
file.write("PySpark\n")
file.write("Azure\n")
file.write("Synapse\n")


##### close the file
file.close()
```

3. append()

It is used to add the new contents at the end of the existing content.

```python
##### open the file in append mode
file = open("data.txt","a")

##### write the content into the file
x = file.write("Data Engineering")
print(x)

##### close the file
file.close()
```

4. with open()

using with open , we can perform all the necessary operations on the file.

```python
with open("data.txt","r") as f:
    ##### read all the contents of the file
    x = f.read()
    print(x)

##### close the file
f.close()
```

##### 5. File pointer

When a file is opened, Python maintains a current position called the file pointer.

tell()

Returns the current position

```python
##### open the file in read mode
file = open("data.txt","r")
##### tell me the current position of the file
x = file.tell()
print(x)
##### close the file
file.close()
```

6. seek()

Moves the file pointer

```python
##### open the file in read mode
file = open("data.txt","r")
##### move the file to the fifth position
file.seek(5)
##### read the contents from the fifth position
x = file.read()
print(x)
##### close the file
file.close()
```

22. Why would you use a instead of w?

a stands for append. It just adds the extra content at the end of the existing file. It does not delete the details of the existing file.

whereas w stands for write. It write the contents into the file. If existing contents exists, it deletes the existing contents and writes the new contents in place of it.

```python
##### open the file using write mode
with open("data1.txt","w") as f:
    f.write("New data\n")
```

```python
##### open the file using append mode
with open("data1.txt","a") as f:
    f.write("Python\n")
    f.write("SQL\n")
    f.write("PySpark\n")
    f.write("Azure\n")
    f.write("DataBricks\n")
f.close()
```

26. How can you return the pointer to the beginning?

```python
##### open the file in read mode
with open("data1.txt","r") as f:
    ##### read the entire file
    x=f.read()
    print(x)
    ##### get the location of the file pointer
    x=f.tell()
    print(x)
    ##### move the cursor pointer to the beginning of the file
    f.seek(0)
    ##### get the current location of the file pointer
    x=f.tell()
    print(x)
```

40. Demonstrate the complete File Handling workflow

Problem

Perform the following:

Create/write a file.

Read it.

Append new content.

Read it again.

Move the pointer.

Read from that position.

```python
with open("data.txt", "w") as file:
    file.write("Python File Handling\n")
    file.write("Reading and Writing")

with open("data.txt", "r") as file:
    data = file.read()

print(data)

with open("data.txt", "a") as file:
    file.write("\nAppending New Data")

with open("data.txt", "r") as file:

    file.seek(5)

    data = file.read()

print(data)
```
