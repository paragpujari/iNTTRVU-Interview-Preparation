##### Function parameter

It is the parameter that is declared and defined inside the function definition.

```python
##### Print "Hello World" using function parameter

##### define the function
def print_data(x):
    print(x)

#### call the function
print_data("Hello World")
```

##### Function Arguement

It is the arguement or the value that is defined while calling the function.

```python
##### Print "Hello World" using function arguement

##### define the function
def print_res(x):
    print(x + "." + "My name is Parag")
#### call the function
print_res("Hello World")
```

##### OBSERVATIONS:

1.  x -----------> Function parameter. It is the parameter defined while declaring the function.

2.  "Hello World" ------> Function arguement. The value defined while calling the function.

##### Return statement

It is the statement that sends the value back to the caller.

```python
##### Add two numbers using return statement
def add(x,y):
    return(x+y)

result = add(10,20)
print(result)
```

##### OBSERVATIONS:

1.  Here the value of 10 is passed to x and the value of 20 is passed to y in the function definition.

2.  Then the addition of x and y takes place and its result is passed to the function call.

3.  In the function call, its resultant value is then stored back to the result value.

4.  Then the result value is printed.

##### Default Arguements

1.  It is the value that is defined while defining the function definition.

```python
##### Print "Hello" with the username while defining the function

##### function definition
def greet(user="Nippon"):
    print("Hello" + " " + user)

##### function call
greet()
##### function call with parameter
greet("Parag")
```

##### OBSERVATIONS:

1.  Here user="Nippon" is the defauly arguement.

2.  It is so because, in the first function call, we can see that greet() ---> there is no value or arguement passed while calling the function.

3.  So by default, the value of user="Nippon" in the function definition is considered as the value for performing the operations inside the function definition.

##### Positional Arguements

1.  These are the arguements that are defined based on their positions and even with respect to the positions, the values from the function call is passed to the parameters in the function definition.

```python
##### Print name and age based on the positional arguements

##### function definition
def print_employee_details(name,age):
    print(f"Employee name is {name}")
    print(f"Employee age  is {age}")
##### function call
print_employee_details("Parag",31)
```

##### OBSERVATIONS:

1.  "Parag" value is defined first in function call and is passed to the name in function definition as it is there defined first.

2.  31 age value is defined second in function call and then it is passed to the second parameter "age" in the function definition.

##### Keyword Arguements

1.  These are the arguements that are passed using the parameter name.

```python
##### Print employee name and age using keyword arguement

##### function definition
def print_details_empdata(name,age):
    print(f"Hi, my name is {name}. My age is {age}.")
##### function call
print_details_empdata(age=31,name="Parag")
```

##### OBSERVATIONS:

1.  Based on the parameter names in the function call, the values from the function call is passed to the respective parameters in the function definition.

##### *args

It allows the function to accept multiple positional arguements.

```python
##### Get the sum of all the numbers using *args

#### function definition
def sum_numbers(*args):
    s=0
    for i in args:
        s = s + i
    return(s)

#### function call
ans = sum_numbers(10,20,30,40,50)
print(f'Resultant value is {ans}')
```

##### OBSERVATIONS:

1.  Here we are passing the values from 10...50 to *args and then calculating the sum of all the five numbers using args.

2.  Then the result of sum is sent back to the function call where it is printed.

##### **kwargs

It accepts multiple keyword arguements.

```python
#### print the student details using keyword arguements

#### function definition
def print_student_details(**kwargs):
    print(kwargs)
#### function call
print_student_details(name="Parag",gender="Male",age=31)
```

##### Returning Multiple Values

A Python function can return multiple values

```python
##### perform the calcutaion and return addition and substraction of two numbers

##### function definition
def calculation(num1,num2):
    res1 = num1 + num2
    res2 = num1 - num2 
    return res1,res2

##### function call
x = 10
y = 7
addition,substraction = calculation(x,y)
print(addition)
print(substraction)
```

34. Create a function that calculates the average of numbers passed using *args

```python
##### define the function
def averages(*args):
    s=0
    for i in args:
        s=s+i
    l=len(args)
    avg = s/l
    return(avg)
##### call the function
ans = averages(10,20,30)
print(f"The average of the numbers is {ans}")
```

35. Create a recursive function to calculate the sum from 1 to n

```python
##### define the function
def rec_sum(x):
    if(x<=0):
        return(0)
    else:
        return(x + rec_sum(x-1))
##### call the function
ans = rec_sum(10)
print(f"The sum of all the numbers from 1 to n is {ans}")
```

40. Create a function that accepts a number and uses another function to calculate its square and cube
