##### Exception

1.  An exception can be defined as the error or the unwanted situation that occurs during the entire execution of the program.

##### Exception Handling

1.  It refers to the technique of handling the errors that occur during the entire execution of the program.

##### Division by Zero

```python
# print(10/0)  ##### ZeroDivisionError: division by zero has occured
```

##### Terminologies:

1. try ----------->  It is used to throw an error to handle it if the error has occured.

2. except --------->  It is used to catch and handle the error if the error has occured.

```python
##### Division by Zero
try:
    print(10/0)
except ZeroDivisionError:
    print("Division By Zero exception has occured")
```

##### Multiple Except blocks

1. For a try statement, we can have multiple except statements

```python
try:
    print(10/0)
except ZeroDivisionError:
    print("Zero Division Error has occured")
except TypeError:
    print("Type Error has occured")
```

##### Finally

Finally statement gets executed whether the exception has occured or not.

```python
try:
    print(10/0)
except ZeroDivisionError as e1:
    print("Zero Division Exception has occured")
except ValueError as e2:
    print("Value Error has occured")
except TypeError as e3:
    print("Invalid Type Error has occured")
finally:
    print("Execution has been successfully completed")
```

##### raise

raise is used to raise an exception.

```python
try:
    raise ValueError("Invalid error has occured")
except ValueError as e:
    print(e)
```

| Exception           | Meaning                         |
| ------------------- | ------------------------------- |
| `ZeroDivisionError` | Division by zero                |
| `ValueError`        | Correct type but invalid value  |
| `TypeError`         | Invalid operation between types |
| `IndexError`        | Invalid index                   |
| `KeyError`          | Missing dictionary key          |
| `NameError`         | Variable/name not defined       |
| `AttributeError`    | Invalid attribute/method        |
| `FileNotFoundError` | File doesn't exist              |
| `ImportError`       | Import problem                  |
| `Exception`         | General exception base class    |
