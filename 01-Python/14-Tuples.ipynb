##### Tuples

1.  A Tuple can be defined as the ordered, immutable collection of elements in Python.

2.  The tuples can have duplicate elements also.

3.  It supports indexing and have elements starting with the index position from 0 till (n-1).

4.  The elements in the tuples are enclosed within ().

5.  It supports slicing 

```python
##### empty tuple
tuples = ()
print(tuples)
print(type(tuples))
```

```python
tuples = (1)
print(tuples)
print(type(tuples))
```

```python
##### tuple with one element
tuples = (1,)
print(tuples)
print(type(tuples))
```

```python
###### tuples with elements of same type
tuples = (1,2,3,4,5)
print(tuples)
print(type(tuples))
```

```python
##### tuples with elements of different types
tuples = (1,2.5,"Hello",True)
print(tuples)
print(type(tuples))
```

##### Tuple Indexing

1.  The indexing of the tuple begins from element 0 and ends till (n-1)

```python
tuples = (1,2,3,4,5)

##### Positive indexing
print(tuples[0])
print(tuples[1])
print(tuples[2])
print(tuples[3])
print(tuples[4])
```

```python
##### Negative Indexing
print(tuples[-1])
print(tuples[-2])
print(tuples[-3])
print(tuples[-4])
print(tuples[-5])
```

##### Tuple Slicing

It refers to the technique of extarcting a portion of tuple from the entire tuple

```python
tuples = (1,2,3,4,5)
#### middle 3 elements
print(tuples[1:4])
print(tuples[:3])  ##### first 3 elements
print(tuples[2:])  ##### last 3 elements
print(tuples[::-2]) ##### reverse order with step size 2
```

##### Tuple Immutability

1.  We cannot change an element in the tuple

```python
tuples = (1,2,3,4,5)

#tuples[2] = 10  ##### 'tuple' object does not support item assignment
print(tuples)
```

##### Tuple Packing

Multiple values can be packed into a tuple

```python
numbers = 10, 20, 30

print(numbers)

print(type(numbers))
```

##### Tuple Unpacking

Extracting each and every element from the tuple

```python
##### Tuple packing
numbers = 10,20,30,40,50
print(numbers)
##### Tuple unpacking
a,b,c,d,e = numbers
print(a)
print(b)
print(c)
print(d)
print(e)
```

##### Tuple Methods

1. count() --------->  It is used to count the occurances of an element in the tuple

```python
numbers = (10,20,10,30,10,40,50)
print(f"No of occurances of element 10 in the tuple is {numbers.count(10)}")
```

2. index() ----------> It returns the position of the first occurrence

```python
numbers = (10,20,10,30,10,40,50)
print(f"Index position of element 10 in the tuple is {numbers.index(10)}")
```

##### Tuple concatenation

It combines two tuples

```python
tup1 = (1,2,3)
tup2 = (4,5,6)

mytup = tup1 + tup2
print(mytup)
```

##### Tuple Repetation

It repeats the times for 'n' number of times

```python
mytup = (1,2,3)
print(mytup * 3)
```

##### Membership in Tuple

```python
mytuple = (1,2,3,4,5)
print(1 in mytuple)
print(10 in mytuple)
```

```python
print(2 not in mytuple)
print(13 not in mytuple)
```
