##### Loops

1.  It is used to iterate over the number of times and produce the statements unless and until a particular condition is satisfied.

2.  It is used to execute a particular block of code unless a particular condition is satisfied.



##### Types of loops

1.  for loop

```python
for i in range(11):
    print(i)
```

2. while loop

```python
i=1
while(i<6):
    print(i)
    i=i+1
```

##### Intermediate Level Interview Questions

11. Calculate the sum from 1 to 10

```python
s=0
i=1
while(i<=10):
    s=s+i
    i=i+1
print(f'Sum of all the numbers from 1....10 is {s}')
```

13. Count how many numbers exist from 1 to 100

```python
c=0
i=1
while(i<=100):
    c=c+1
    i=i+1
print(f'Total numbers from 1....100 is {c}')
```

14. Find the largest number using a loop

```python
numbers = [10, 25, 7, 40, 15]


largest=0
for i in numbers:
    if(i>largest):
        largest=i
print(f'Largest element in the array is {largest}')
```

15. Find the smallest number using a loop

```python
numbers = [10, 25, 7, 40, 15]

smallest = numbers[0]

for i in numbers:
    if(i < smallest):
        smallest = i
print(f'Smallest element in the array is {smallest}')   
```

16. Count numbers greater than 10

```python
for i in range(1, 10):
    
    if i == 5:
        break

    print(i)
```

##### OBSERVATIONS:

```python
for i in range(1, 6):
    
    if i == 3:
        continue

    print(i)
```

##### OBSERVATIONS:

1.  When continue statement executes, then it will skip the current iteration of the loop.
