##### Strings

1.  A string can be defined as the sequence of characters enclosed within single or double quotes.

```python
name = "Parag"

city = "Hyderabad"

print(f"Name: {name}")
print(f"City: {city}")
```

##### Properties of Strings

1.  Strings are immutable. We cannot change any character in the original string.

To change or perform any modifications, just create a new string and perform modifications there

```python
language = "Python"

print(language)

print('J' + language[1:])
```

```python
##### Obtain the length of the string

name = "Parag"

l = len(name)

print(f'Length of the string is {l}')
```

```python
##### Obtain the last character in the string 
last_ch = name[-1]
print(f'Last character in the string is {last_ch}')
```

```python
##### Reverse the string
x = 'Python'
reverse_x = x[::-1]
print(f'Reversed string is {reverse_x}')
```

21. Find the first character that appears twice

```python
x = 'programming'

freq={}

for i in x:
    if(i in freq):
        freq[i] = freq[i] + 1
    else:
        freq[i] = 1

for x,y in freq.items():
    if(y==2):
        print(x)
        break
```

39. Compress consecutive characters

```python
x = "aaabbc"
c = 1
res = ""

for i in range(1,len(x)):
    if(x[i] == x[i-1]):
        c = c + 1
    else:
        res = res + x[i-1] + str(c)
        c = 1
res = res + x[i-1]+str(c)
print(res)
```

40. Find the longest palindromic substring

```python
s = "babad"
longest = ""

for i in range(len(s)):
    for j in range(i + 1, len(s) + 1):
        current = s[i:j]

        if current == current[::-1]:
            if len(current) > len(longest):
                longest = current

print(longest)
```
