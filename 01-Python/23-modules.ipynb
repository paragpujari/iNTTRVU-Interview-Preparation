##### Module

1.  A module is a python file that contains the python code which can be reused in some other python programs.

##### Importance of module

1.  module helps in organizing the code.

2.  It helps in reusing the same set of codes.

3.  It prevents us from writing the same set of codes repetatively for multiple number of times.

4.  It improves the inbuilt functionality of the program


##### Types of modules

1. Inbuilt modules

These modules are already made by the system.

```python
import math

x = math.sqrt(16)

print(x)
```

```python
import random

x = random.randint(1,100)

print(x)
```

```python
import datetime
```

2. User defined modules

These are the modules that are made by the user.

##### Excercises on Module

1. find the square root of 16

```python
import math
x = math.sqrt(16)
print(x)
```

2. Import a Specific Item called sqrt from math

```python
from math import sqrt
x = sqrt(16)
print(x)
```

3. Import Multiple Items sqrt, pow from math

```python
from math import sqrt,pow

##### Find the square root of the number
x = sqrt(16)
print(f"Square root of 16 is {x}")

##### Get the power of the number
x = pow(2,3)
print(f"The power of 2 to 3 is {x}")
```

4. Import Module Using an Alias

```python
import math as m
x = m.sqrt(16)
print(x)

import math as m
x = m.pow(2,3)
print(x)
```

5. Module Using a User-Defined Module

```python
import calculator
print(calculator.result)  ##### Access the result from the calculator file
```
