#### Operators

1.  An operator can be defined as the symbols with which we can perform operations on the values.

2.  It is basically used for performing the mathematical operations.

##### Main Operators Categories

| Category   | Operators                  | Purpose                   |
| ---------- | -------------------------- | ------------------------- |
| Arithmetic | `+ - * / // % **`          | Mathematical calculations |
| Comparison | `== != > < >= <=`          | Compare values            |
| Assignment | `= += -= *= /= //= %= **=` | Assign/update values      |
| Logical    | `and or not`               | Combine conditions        |
| Bitwise    | `& \| ^ ~ << >>`           | Work at bit level         |
| Membership | `in`, `not in`             | Check membership          |
| Identity   | `is`, `is not`             | Check object identity     |

##### Types of Operators

##### 1. Arithmetic Operators

```python
##### performing the arithmetic operations
a = 10
b = 3

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
print(a ** b)
```

##### 2. Comparision Operators

```python
##### performing the arithmetic operations
a = 10
b = 3

print(a > b)
print(a < b)
print(a == b)
print(a != b)
print(a >= b)
print(a <= b)
```

##### 3. Assignment Operators

```python
##### performing the assignment operations
a = 10
a += 3

b = 10
b -= 3

c = 10
c *= 3

d = 10
d /= 3

e = 10
e %= 3

f = 10
f //= 3

g = 10 
g **=3

print(a)
print(b)
print(c)
print(d)
print(e)
print(f)
print(g)
```

##### 4. Logical Operators

```python
print((10 > 5) and (8 > 6))

print((2 > 5) or (10 > 13))

print(not(10>5) or (9>8))
```

##### 5. Bitwise Operators

```python
a = 5
b = 3

##### perform bitwise AND
print(a&b)

##### perform bitwise OR
print(a|b)

##### perform bitwise XOR
print(a^b)

#### perform bitwise NOT
print(~a)

##### perform bitwise left shift
print(a<<1)

##### perform bitwise right shift
print(a>>1)
```

##### 6. Membership Operators

```python
print(3 in [1, 2, 3])

print(5 in [1,2,3,4])

print(4 not in [1,2,3,4])

print(10 not in [1,2,3])
```

##### 7. Identity Operators

```python
a = 10
b = 10

print(a is b)


a = [1,2,3]
b = [1,2,3]

print(a is b)
```

##### Operator Precedence

Python follows precedence rules.



```python
result = 10 + 5 * 2
print(result)
```

```python
##### Using parenthesis
result = (10 + 5)*2
print(result)
```
