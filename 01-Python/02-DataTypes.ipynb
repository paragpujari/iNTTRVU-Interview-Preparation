##### Data Types

1.  A Data Type can be defined as the type assigned to each and every element used in the Python Program.

##### Main Inbuilt Data Types

| Category | Data Types                |
| -------- | ------------------------- |
| Numeric  | `int`, `float`, `complex` |
| Boolean  | `bool`                    |
| Text     | `str`                     |
| Sequence | `list`, `tuple`, `range`  |
| Mapping  | `dict`                    |
| Set      | `set`, `frozenset`        |
| Binary   | `bytes`, `bytearray`      |
| Special  | `NoneType`                |

37. Find the unique values in two lists using set data types

```python
a = [1, 2, 3, 4]
b = [3, 4, 5, 6]

#### Get the unique values in two lists using sets
ans = set(a) | set(b)
print(ans)
```

38. Find common values between two lists using sets

```python
a = [1, 2, 3, 4]
b = [3, 4, 5, 6]
##### Get the common values in two lists using sets
a = set(a)
b = set(b)
ans = a.intersection(b)
print(ans)
print(type(ans))
```

39. Create a dictionary from two lists

```python
keys = ["name", "age", "city"]
values = ["Parag", 30, "Hyderabad"]

##### Create the dictionary from the two lists
ans = dict(zip(keys,values))
print(ans)
print(type(ans))
```
