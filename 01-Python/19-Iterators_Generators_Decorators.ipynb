##### Iterators

1.  An iterator is an object that allows to access one element at a time.

##### Teo important methods of iterators

1.  iter() -----> It is an object through which we can access one element at a time.

2.  next() -----> It helps in fetching the next value.

```python
##### iterator
numbers = [1,2,3,4,5]

##### using iter, it helps in acheiving one value one after the other
x = iter(numbers)
print(x)

##### fetch one element one by one
print(f"Elements fetched one by one is {next(x)}")
print(f"Elements fetched one by one is {next(x)}")
print(f"Elements fetched one by one is {next(x)}")
print(f"Elements fetched one by one is {next(x)}")
print(f"Elements fetched one by one is {next(x)}")
```

##### Generator

A Generator is an object that produces one value once at a time instead of producing everything at once.

It has a keyword named yield.

```python
##### create a function definition
def numbers():
    yield 1
    yield 2
    yield 3

##### create a function call
gen = numbers()
print(next(gen))
print(next(gen))
print(next(gen))
```

##### Decorators

A decorator modifies the behaviour of the code without changing the actual meaning of the code.

```python
def decorator(function):
    def wrapper():
        print("Before")
        function()
        print("After")

    return(wrapper)

##### define the function call
@decorator
def greet():
    print("Hello World")


greet()
```

30. Explain the execution order of this code

```python
def A(function):
    
    def wrapper():
        print("A")
        function()

    return wrapper


def B(function):

    def wrapper():
        print("B")
        function()

    return wrapper


@A
@B
def test():
    print("Test")


test()
```

##### OBSERVATIONS:

The decoration happens conceptually as:

test = A(B(test))

Therefore the outer decorator A executes first when test() is called

40. Create a decorator around a generator function

```python
def decorator(function):
    
    def wrapper():

        print("Generator started")

        gen = function()

        for value in gen:
            print(value)

        print("Generator finished")

    return wrapper


@decorator
def numbers():

    yield 10
    yield 20
    yield 30


numbers()
```
