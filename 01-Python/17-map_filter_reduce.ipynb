##### map()

1.  map() -------->  It is used to apply same operation or function to each and every element.

##### Syntax

map(function, iterable)

```python
numbers = [1,2,3,4,5]

##### Squares of each and every element in the list
res = list(map(lambda x:x**2,numbers))

print(res)
```

filter()

2. filter() -------> It is used when you want to keep only elements satisfying a condition

##### Syntax

filter(function,iterable)

```python
numbers = [1,2,3,4,5]

##### filter all the numbers greater than 3
res = list(filter(lambda x:x>3,numbers))
print(res)
```

reduce()

3. reduce() ----------> It is used when you want to combine multiple values into one final value

##### Syntax

reduce(function,iterable)

```python
numbers = [1,2,3,4,5,6]

##### Find out the product of all the numbers in the list
from functools import reduce

res = reduce(lambda x,y:x*y,numbers)

print(res)
```
