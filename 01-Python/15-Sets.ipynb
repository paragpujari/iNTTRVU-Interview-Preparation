##### Sets

1.  A set can be defined as the unordered collection of unique elements.

2.  It is always enclosed within {}.

3.  We can change the elements in the sets. So it is mutuable.

4.  We cannot perform indexing on the sets.

5.  Traversing is possible in the sets.

6.  It can have the elements of same or different types.

```python
##### empty set
sets = set({})
print(sets)
print(type(sets))
```

##### OBSERVATIONS:

1.  For an empty set, we are declaring it using set() because by default, the empty {} will be considered as dictionary.


```python
##### sets with elements
sets = {1,2,3,4,5}
print(sets)
print(type(sets))
```

```python
##### sets with different elements
sets = {1,2.5,"Hello",(1,2,3),True}
print(sets)
print(type(sets))
```

```python
##### Sets can have only unique elements. No duplicates will be allowed in the sets.
sets = {1,1,1,2,2,3,3,3,4}
print(sets)
print(type(sets))
```

##### Basic Operations in Sets

1. add() ---------> It adds the elements in the set.

```python
sets = {10,20,30,40}
#### Add element 50 in the set
sets.add(50)
print(sets)
```

2. remove() -------> It deletes an element from the set. If element does not exists , it throws an error.

```python
sets = {10,20,30,40,50}
##### Remove element 50 from the set
sets.remove(50)
print(sets)
```

```python
sets = {10,20,30,40,50}
##### Remove element 60 from the set
# sets.remove(60) # KeyError: 60, as element 60 does not exists
print(sets)
```

3. discard() ---------> It deletes an element from the sets. If element does not exists, nothing happens.

```python
sets = {10,20,30,40,50}
##### Remove element 60 from the set
sets.discard(60) 
print(sets)
```

##### OBSERVATIONS:

1. Element 60 doesnot exists in the sets. But here we are trying to remove 60 from the sets. Nothing happens as discard does not affect the element that does not exists.

4. clear() ----------> It removes all the elements from the sets

```python
sets = {10,20,30,40,50}
##### remove all the element from the sets
sets.clear()
print(sets)
```

5. len() ----------> It determines the total number of elements present in the sets.

```python
sets = {10,20,30,40,50}
##### Obtain the length of the sets
l = len(sets)
print(f"Length of the sets is {l}")
```

6. in ----------> It determines whether the element exists in the sets or not.

```python
sets = {10,20,30,40,50}
print(10 in sets)
```

##### Set Mathematical Operations

1. Union

It combines the elements from both the sets

```python
a = {1,2,3}
b = {4,5,6}

print(a|b)

print(a.union(b))
```

2. Intersection

It takes the common elements from both the sets.

```python
a = {1,2,3}
b = {2,3,4}

print(a&b)
print(a.intersection(b))
```

3. difference

It considers the elements present in one set but not in the other.

```python
a = {1,2,3}
b = {2,3,4}

print(a-b)
print(a.difference(b))


print(b-a)
print(b.difference(a))
```

4. Symmetric difference

It considers the elements that are present in either of the sets.

```python
a = {1,2,3}
b = {2,3,4}

print(a^b)
```

##### Set relationships

1. issubset()

```python
a = {1, 2}
b = {1, 2, 3}

print(a.issubset(b))
```

2. issuperset()

```python
a = {1,2,3}
b = {1,2}

print(a.issuperset(b))
```

3. Disjoint

Two sets are disjoint when they have no common elements

```python
a = {1,2}
b = {3,4}

##### To determine whether both sets are different
print(a.isdisjoint(b))
```
