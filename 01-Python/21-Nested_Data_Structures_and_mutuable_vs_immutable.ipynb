##### Nested Data Structures

    When one data structure is present inside another data structure, then it is known as nested data structure.

```python
##### Nested lists
data = [[1,2],[3,4]]

print(data)

##### Nested dictionary
data = {
    "student" : {
        "name"  :   "Parag" ,
        "age"   :   31
    }
}

print(data)

##### Nested tuples
data = ((1,2),(3,4,5))

print(data)
```

```python
##### Mutuable  ---------> Elements in the iterable can be changed
numbers = [1,20,3,4,5]
##### change the element at index 1
numbers[1] = 2
print(numbers)
```

```python
##### Immutable --------> Elements in the tuple cannot be changed
numbers = (1,2,3,4,5)
##### change the element at index 1
#numbers[1] = 20  ##### 'tuple' object does not support item assignment
print(numbers)
```

##### Basic Level Interview Questions

1. Access a Value from a Nested List. Print 40

```python
data = [[10, 20], [30, 40]]

print(data[1][1])
```

2. Modify a Nested List. Change 30 to 300

```python
data = [[10, 20], [30, 40]]

##### Change 30 to 300
data[1][0]=300
print(data)
```

3. Identify Mutable and Immutable Nested Objects. Which inner structure can be modified?

```python
data = [[10, 20], (30, 40)]

##### Mutuable nested object
print(f"Mutuable nested object is {data[0]}")
```

##### OBSERVATIONS:

1.  [10,20] is a mutuable nested object as it is the list and the elements in the list can be changed at any index position. So the list nested element can be modified.

```python
##### Immutable nested object
print(f"Immutable nested object is {data[1]}")
```

##### OBSERVATIONS:

1. (30,40) is an immutable nested object as it is a tuple and the elements in the tuple cannot be changed and it does not support item assignment.

##### Intermediate Level Interview Questions

4. Modify a Nested Dictionary. Change the age to 26.

```python
data = {
    "student": {
        "name": "Rahul",
        "age": 25
    }
}

data["student"]["age"]=26
print(data)
```

5. Nested List Inside a Tuple. Can we change 10 to 100?

```python
data = ([10, 20], 30)
##### Change 10 to 100
data[0][0]=100
print(data)
```

##### OBSERVATIONS:

1.  Yes we can change 10 to 100 inside the nested structure. This is because the element 10 is present inside the list in a nested structure.

6. Nested Dictionary Containing a List. Change 70 to 75

```python
data = {
    "marks": [80, 90, 70]
}
```

```python
##### Change 70 to 75 in the dictionary
data["marks"][2]=75

print(data)
```

10. Modify Multiple Nested Values. Change 10->100, 40->400

```python
data = (
    [10, 20],
    [30, 40]
)

data[0][1]=200
data[1][1]=400

print(data)
```
