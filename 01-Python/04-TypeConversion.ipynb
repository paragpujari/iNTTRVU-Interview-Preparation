##### Type Conversion

1.  A Type Conversion is a process of converting the data of one type into another type.

##### Loss of precision

1.  The Loss of precision occurs when we want to convert one data type to another and the converted value does not represent the original value in the same original format.

2.  Loss of precision occurs when we want to convert from higher -----> lower data type

3.  When we convert the data from lower -----> higher data type, there is no loss of precision.

##### No loss of precision

smaller-------->higher data type

int------------>Float

```python
x = 10
print(f'Original value is {x}')
##### integer -----> float
ans = float(x)
print(f'Converted value is {ans}')
```

##### OBSERVATIONS:

1.  Here as we are converting from lower to higher data type(int-------->float), there results in no loss of precision.

##### Loss of precision

larger-------------->smaller data type

float---------------->int

```python
x = 10.75
print(f'Original value is {x}')
#####float--------->int
ans = int(x)
print(ans)
```

##### OBSERVATIONS:

1.  Here as we are converting the value from higher to smaller data type(float--->int), there results in the loss of precision.

##### Types of Type Conversion

1.  Implicit Type Conversion

    It is a type of type conversion where we are automatically converting the value of one type to another.

```python
x = 10
y = 0.75

res = x+y
print(res)
print(type(res))
```

##### OBSERVATIONS:

1.  Initially we have x as 10 which is integer. Then we are adding 0.75 to it and then storing the result in res.

2.  Then the res value is in float type.

2. Explicit Type Conversion

Here the programmer needs to manually convert the type of value into another type.

```python
x = "10.75"

##### Convert it into float type
res = float(x)
print(res)
print(type(res))
```

```python
x = "100"

##### Convert it into int type
res = int(x)
print(res)
print(type(res))
```

```python
x = 100

##### Convert it into string type
res = str(x)
print(res)
print(type(res))
```

```python
x = 1

##### Convert it into bool type
res = bool(x)
print(res)
print(type(res))
```
