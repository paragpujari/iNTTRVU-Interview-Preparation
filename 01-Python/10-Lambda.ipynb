##### Lambda Function

1.  A lambda function is an anonymous function written in a single expression.

```python
##### Square of a number
ans = lambda x:x**2
print(f"The square of the number is {ans(5)}")
```

##### Basic Level Interview Questions
1. Create a lambda to add 5 to a number

```python
ans = lambda x:x+5
print(f"The answer is {ans(10)}")
```

2. Create a lambda to double a number

```python
ans = lambda x:x*2
print(f"The answer is {ans(5)}")
```

5. Create a lambda to subtract two numbers
