##### Lists

1.  A list can be defined as the ordered, mutuable collection of elements in python.

2.  It can be considered as the collection of elements within [].

3.  The indexing of lists begins from 0 and ends till (n-1).

4.  The elements in the lists can be changed at any position.

5.  It can have elements of same or different type

##### Creation of lists

```python
##### empty lists
numbers = []
print(numbers)
print(type(numbers))
```

```python
##### lists with elements of same type
lists = [1,2,3,4,5]
print(lists)
print(type(lists))
```

```python
##### lists with elements of different type
lists = [1,2,True,(2+3j),'Parag',20.35]
print(lists)
print(type(lists))
```

```python
###### Lists can have duplicates
lists = [10,10,20,20,30,40,50]
print(lists)
print(type(lists))
```

##### Indexing

1. The lists start with an index position 0 and ends till (n-1)

```python
numbers = [10, 20, 30, 40, 50]

print('Positive Indexing')
##### get the elements in ascending
print(numbers[0])
print(numbers[1])
print(numbers[2])
print(numbers[3])
print(numbers[4])
print('--------------------------------------------------------------------------------')
print('Negative Indexing')
##### get the elements in descending
print(numbers[-1])
print(numbers[-2])
print(numbers[-3])
print(numbers[-4])
print(numbers[-5])
```

##### Slicing

It refers to the process of extracting the portion of list from the entire list.

```python
numbers = [10, 20, 30, 40, 50]
print(numbers[1:4])

##### Get all the even position numbers
print(numbers[0:5:2])

##### print the numbers in reverse order
print(numbers[::-1])
```

##### Modifying Lists elements

1.  Lists are mutuable.

```python
numbers = [10,15,30]

##### Change the element of tyhe list at index position 1
numbers[1] = 20
print(numbers)
```

##### Addition of element in the list

1. append() --------> It adds the element at the end of the list.

```python
numbers = [10,20,30,40]
print(f"Elements before appending the list is:{numbers}")
##### Add 50 to the end of the list
numbers.append(50)
print(f"Elements after appending the list is:{numbers}")
```

2. insert() --------> It adds the element at the specific index position.

```python
numbers = [10,30,40]
print(f"Elements before inserting the element in the list is:{numbers}")
##### insert element 20 at index 1
numbers.insert(1,20)
print(f"Elements after inserting the element in the list is:{numbers}")
```

3. extend() -----> It adds one list at the end of another list

```python
a = [10,20,30]
b = [40,50]
print(a)
a.extend(b)

print(a)
```

##### Removal of elements from the list

1. remove() ----------> removes a particular element from the list.

```python
numbers = [10,20,30,40]
print(numbers)
##### remove element 20 from the list
numbers.remove(20)
print(numbers)
```

2. pop() --------------> removes the element based on the index position.

```python
numbers = [10,20,30,40]
print(numbers)
##### Remove element present at index 1
numbers.pop(1)
print(numbers)
```

3. clear() ---------> removes all the elements from the list

```python
numbers = [10,20,30,40]
print(numbers)
##### Removes all the elements from the list
numbers.clear()
print(numbers)
```

##### List Searching

```python
##### index
numbers = [10, 20, 30, 40]

print(numbers.index(20))  ##### It obtains the index position of the element in the list
```

```python
##### count
numbers = [10,20,30,40]
print(numbers.count(20))
```

##### Sorting

It is used to arrange the elements of the list in ascending or descending order

```python
lists = [10,30,20,50,40,60,70]
print(lists)
##### Sort the lists in ascending order
lists.sort()
print(lists)
```

```python
lists = [10,30,20,50,40,60,70]
print(lists)
##### Sort the lists in descending order
lists.sort(reverse=True)
print(lists)
```

##### Reverse

```python
numbers = [10,20,30,40,50]
print(numbers)

##### Reverse the list
numbers.reverse()
print(numbers)
```

##### Copying the list

```python
numbers = [10,20,30,40,50]
print(numbers)

##### Copy the numbers to another location
numbers1 = numbers.copy()
print(numbers1)
```
