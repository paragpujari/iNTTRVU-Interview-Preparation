##### Dictionary

1.  It is a data structure in Python that is used to store the element in key-value pair.


##### Properties of Dictionary

1.  It stores the elemenst in key-value pair.

2.  Keys in the dictionaries must be unique.

3.  Values in the dictionaries can have duplicate values.

4.  Keys in the dictionary can be hashable.

5.  Dictionaries are mutuable. We can add or remove the element from the dictionary.

6.  Dictionaries are ordered. It reserves the order in which the items in the dictionary are placed.

7.  It can have ssame or different types of elements.

8.  The values in the dictionaries can be accessed by means of key.

27. Find the difference between the largest and smallest dictionary values

```python
data = {
    "A": 40,
    "B": 90,
    "C": 60,
    "D": 20
}


##### Get the largest value in the dictionary
largest = None
for y in data.values():
    if(largest is None or y > largest):
        largest = y
print(largest)

##### Get the smallest value in the dictionary
smallest_no = None
for y in data.values():
    if(smallest_no is None or y < smallest_no):
        smallest_no = y 
print(smallest_no)

##### Get the difference between the largest and smallest dictionary values
diff_values = largest - smallest_no
print(f"The difference between the largest and smallest dictionary values is {diff_values}")
```

40. Create a reversed dictionary

```python
data = {
    "A": 100,
    "B": 200,
    "C": 300
}

datas = {}
##### To create a reversed dictionary
for x,y in data.items():
    print(y,":",x)
    datas[y] = x
print(datas)

##### Clear the original dictionary
data.clear()
print(data)

##### Copy the contents from the modified to the original dictionary
data = datas.copy()
print(data)

del datas
```
