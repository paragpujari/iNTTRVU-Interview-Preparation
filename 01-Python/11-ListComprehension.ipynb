##### List Comprehension

List comprehension is a short way of creating a list from an existing iterable


##### Syntax

answer = [expression for item in iterable if condition]

```python
##### Given a list, find out the square of all the elements in the list
numbers = [1, 2, 3, 4, 5]

ans = [i**2 for i in numbers if(i!=0)]

print(ans)
```
