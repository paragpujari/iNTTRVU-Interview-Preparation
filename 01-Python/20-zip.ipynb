##### zip()

1.  zip() combines elements from two or more iterables position by position.

2.  But it restuns the result in form of object.

3.  To show the actual result, we need to apply list above the zip() function

##### Syntax:

zip(iterable1, iterable2,....)

```python
names = ["A","B","C"]
marks = [80,70,90]

result = list(zip(names,marks))

print(result)
```

##### Properties of zip

1.  zip combines the elements from two or more iterables position by position and returns an object

2.  zip() stops the execution when the sortest iterable ends.

```python
names = ['A','B','C']
marks = [80,90]

result = list(zip(names,marks))
print(result)
```

##### Basic Level Interview Questions

1. Combine Two Lists Using zip()

```python
names = ["Amit", "Rahul", "Priya"]
ages = [25, 28, 24]

##### zip names with ages
ans = list(zip(names,ages))

print(ans)
```

2. Combine the following two lists using zip()

```python
a = [10, 20, 30]
b = [1, 2, 3]

##### zip a and b
ans = list(zip(a,b))

print(ans)
```

3. What Happens When Lists Have Different Lengths?

```python
a = [10, 20, 30, 40]
b = [1, 2]

ans = list(zip(a,b))
print(ans)
```

##### OBSERVATIONS:

1.  zip stops the execution where ever the shortest iterable ends.

##### Intermediate Level Interview Questions

6. Create a Dictionary Using zip()

```python
names = ["Amit", "Rahul", "Priya"]
marks = [80, 90, 85]

##### zip names with marks
ans = dict(zip(names,marks))

print(ans)
```

7. Zip and Print Each Pair

```python
names = ["A", "B", "C"]
scores = [90, 80, 70]

##### zip names with scores
ans = dict(zip(names,scores))

print(ans)

for x,y in ans.items():
    print(x, ' ',y)
```

9. Find the Number of Pairs Created by zip()

```python
a = [1, 2, 3, 4, 5]
b = [10, 20, 30]

##### zip a and b
ans = list(zip(a,b))
print(ans)

##### Determine the total number of pairs formed using zip
print(f"The total number of pairs formed using zip is {len(ans)}")
```

##### Analytical Based Interview Questions

11. Why Does zip() Ignore the Remaining Elements?

```python
employees = ["A", "B", "C", "D"]
salaries = [50000, 60000]

##### zip employees with salaries
ans = list(zip(employees,salaries))

print(ans)
```

##### OBSERVATIONS:

1. zip() Ignore the Remaining Elements because it stops the execution once the length of the shortest iterable is over.

12. What Happens When One List Is Empty?

```python
names = []
marks = [80, 90, 70]

##### zip names with marks

ans = list(zip(names,marks))

print(ans)
```

##### OBSERVATIONS:

1.  When one of the list is empty, if zipping takes place with one of the empty lists, then the entire result will be empty.

13. What Happens When zip() Receives Three Iterables?

```python
a = [1, 2, 3]
b = [10, 20, 30]
c = [100, 200, 300]

##### perform zipping with three lists
ans = list(zip(a,b,c))

print(ans)
```

##### OBSERVATIONS:

1.  When we zip three iterables, then three pairs of elements are formed.

14. Can zip() Be Used Directly in a for Loop?

```python
employees = ["A", "B", "C"]
departments = ["IT", "HR", "Sales"]

##### zip employees with departments in a for loop
for x,y in zip(employees,departments):
    print(x,' ',y)
```

15. What Happens If zip() Is Converted to a List Twice?

```python
a = [1, 2, 3]
b = [10, 20, 30]

##### zip a with b
ans = zip(a,b)

##### convert zip to list twice
print(list(ans))
print(list(ans))
```

##### OBSERVATIONS:

1.  In zipping elements from two or more iterables are paired position by position and then it is converted into list to get the actual results.

2.  If again list function is applied on the zipped elements, then it will return nothing because all the elements from both the iterables are paired and no more elements are left.
