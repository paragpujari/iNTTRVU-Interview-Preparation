##### Dictionary Comprehension

1.  It is an enhanced way of creating the dictioanries

##### Syntax:

{key:value for item in iterable}

##### Python module

1.  A python module is a simple python file containing the python code.

```python
##### Get the squares of all the numbers and perform it using dictionary comprehension
numbers=[1,2,3,4,5]
ans = {item:item**2 for item in numbers}
print(ans)
```

##### Basic Level Interview Questions

4. Create a dictionary where numbers 1, 2, 3, 4, 5 are keys and their squares are values

```python
numbers = [1,2,3,4,5]

##### Get the squares of the above list and perform it using dictionary comprehension
ans = {x:x**2 for x in numbers if(x!=0)}

print(ans)
```

5. Create a dictionary containing only even numbers from 1 to 6, with each number as both key and value

```python
numbers = [1,2,3,4,5,6]

##### Create the dictionary using dictionary comprehension
ans = {x:x for x in numbers if(x%2==0)}

print(ans)
```

##### Intermediate Level Interview Questions

8. Create a new dictionary where every price is increased by 10%

```python
prices = {
    "Laptop": 50000,
    "Phone": 30000,
    "Tablet": 20000
}

##### Create the dictionary using dictionary comprehension
ans = {x:(y + (0.10*y)) for x,y in prices.items()}

print(ans)
```

9. Create a dictionary containing only sales greater than 30000

```python
sales = {
    "A": 50000,
    "B": 25000,
    "C": 70000,
    "D": 15000
}

##### Create the dictionary using dictionary comprehension
ans = {x:y for x,y in sales.items() if(y>30000)}

print(ans)
```

10. Create a new dictionary where every key is uppercase

```python
data = {
    "name": "Parag",
    "city": "Hyderabad"
}

##### Create the new dictionary using dictionary comprehension
ans = {x.upper():y for x,y in data.items()}

print(ans)
```

11. Create a dictionary containing products whose price is at least 25000

```python
products = {
    "Laptop": 80000,
    "Mouse": 1000,
    "Monitor": 25000,
    "Keyboard": 3000,
    "Server": 150000
}

##### Create the new dictionary using dictionary comprehension
ans = {x:y for x,y in products.items() if(y >= 25000)}

print(ans)
```

12. Create a new dictionary where every price has a 10% discount

```python
prices = {
    "Laptop": 80000,
    "Phone": 40000,
    "Tablet": 20000
}

##### Create the new dictionary using dictionary comprehension where every price has a 10% discount
ans = {x:(y-0.10*y) for x,y in prices.items() if(y!=0)}

print(ans)
```

13. Create a dictionary from two dictionaries

```python
{
    'Amit': 'IT',
    'Rahul': 'HR',
    'Neha': 'Finance'
}
```

```python
employees = {
    "E01": "Amit",
    "E02": "Rahul",
    "E03": "Neha"
}



departments = {
    "E01": "IT",
    "E02": "HR",
    "E03": "Finance"
}

##### Create a dictionary using the dictionary comprehension
ans = {employees[x] : departments[x] for x,y in employees.items() if(x!=0)}

print(ans)
```

14. Create this dictionary using dictionary comprehension

```python
{
    1: {1: 1, 2: 2, 3: 3},
    2: {1: 2, 2: 4, 3: 6},
    3: {1: 3, 2: 6, 3: 9}
}
```

```python
##### Create the above dictionary using dictionary comprehension
ans = {i:{j:(j*i) for j in range(1,4)} for i in range(1,4) if(i!=0)}

print(ans)
```
