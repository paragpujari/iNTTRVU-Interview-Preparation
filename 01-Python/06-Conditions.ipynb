##### Conditional Statements

1.  These are the statements that are used to apply conditions to make certain decisions.

2.  They are used for decision making tasks by applying certain conditions.


##### Types of Conditional Statements

1.   if condition:
         
         statement

```python
age = 18
##### Eligible to vote
if(age >= 18):
    print(f'Eligible to vote')
```

2.   if condition:
        
        statement1
    
    else:
    
        statement2

```python
##### Eligible to vote or not
age = 10
if(age >= 18):
    print(f'Eligible to vote')
else:
    print(f'Not eligible to vote')
```

3. elif statement

if condition1:
    
    statement1

elif condition2:

    statement2

else:

    statement3

```python
##### Check whether the person is minor, adult or senior citizen
age = 25
if(age < 18):
    print('Minor')
elif((age >= 18) and (age <= 60)):
    print('Adult')
else:
    print('Senior Citizen')
```

4. Nested if statement

if statement:

    if statement:

        stmt

```python
age = 25
citizen = True


if(age >= 18):
    if(citizen):
        print("Eligible")
```

LOGICAL CONDITIONS

and

Both conditions must be true.

```python
age = 25
salary = 50000

if age >= 18 and salary >= 30000:
    print("Eligible")
```

or

Any of the conditions must be true.

```python
marks = 90

if marks >= 90 or marks == 100:
    print("Excellent")
```

not

Reverses the condition

```python
available = False

if not available:
    print("Not Available")
```
